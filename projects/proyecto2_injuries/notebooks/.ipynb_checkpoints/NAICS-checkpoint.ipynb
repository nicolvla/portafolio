```python
import pandas as pd
```

```python
osha = pd.read_csv(
    "../data/OSHA-Severe-Injuries-Report-NYC-DATA-csv.csv",
    dtype = {"Primary NAICS":"str"}
)
osha_naics =pd.read_csv(
    "../data/osha_naics_code.csv",
    dtype ={"naics_code":"str"}
)
```

```python
osha.columns
```

```python
osha_naics.columns
```

Make new column with only the sector code:

```python
osha["naics_sector_code"] = osha["Primary NAICS"].apply(lambda x: str(x)[:2])
```

```python
osha["naics_sector_code"]
```

```python
new_data_frame = pd.merge(
    osha,
    osha_naics,
    left_on = "naics_sector_code",
    right_on = "naics_code"
)
```
