```python
import pandas as pd
```

```python
osha_severe_injuries_a_to_m = pd.read_csv("../data/osha_a_to_n.csv.csv")
osha_severe_injuries_n_to_z = pd.read_csv("../data/osha_m_to_z.csv")
naics_database = pd.read_csv("../data/osha_naics_code.csv")
```

```python
all_databases = pd.concat([osha_severe_injuries_a_to_m, osha_severe_injuries_n_to_z])
print(len(all_databases))
```

```python
all_databases["naics_sector_code"] = all_databases["Primary NAICS"].apply( lambda x: str(x)[:2])
```

```python
all_databases.head()
```

```python
naics_database.head()
```

```python
new_database = pd.merge(
                all_databases,
                naics_database,
                left_on = "naics_sector_code",
                right_on = "naics_code",
)
```

```python
new_database.head()
```

```python
osha_new_database = new_database.rename(columns={'naics_description': 'Industry'})
```

```python
osha_new_database.drop(columns='naics_code')
```

```python
osha_new_database.to_json("osha_new_database.json", indent=4)
```

```python
osha_new_database.to_csv('../output/osha_severe_injuries.csv')
```
